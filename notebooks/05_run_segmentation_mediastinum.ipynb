```python
# Prepare the configuration file
import configparser
import json
import logging

test_results_dir = os.path.join(results_dir, "output_mediastinum_segmentation")
os.makedirs(test_results_dir, exist_ok=True)

logging.basicConfig()
logging.getLogger().setLevel(logging.DEBUG)
rads_config = configparser.ConfigParser()
rads_config.add_section('Default')
rads_config.set('Default', 'task', 'mediastinum_diagnosis') # The task has to be set to the correct target
rads_config.set('Default', 'caller', '')
rads_config.add_section('System')
rads_config.set('System', 'gpu_id', "-1")
rads_config.set('System', 'input_folder', os.path.join(patient_dir, "patient-UnitTest3-Mediastinum", "inputs"))
rads_config.set('System', 'output_folder', test_results_dir)
rads_config.set('System', 'model_folder', models_dir)
rads_config.set('System', 'pipeline_filename', os.path.join(pipeline_dir, 'pipeline_mediastinum_segmentation.json'))
rads_config.add_section('Runtime')
rads_config.set('Runtime', 'reconstruction_method', 'thresholding')
rads_config.set('Runtime', 'reconstruction_order', 'resample_first')
rads_config.set('Runtime', 'use_stripped_data', 'False')  # True when working will skull-stripped data
rads_config.set('Runtime', 'use_registered_data', 'False')  # True when working will co-registered input MR scans (e.g., BraTS)

rads_config_filename = os.path.join(test_results_dir, 'rads_config.ini')
with open(rads_config_filename, 'w') as outfile:
    rads_config.write(outfile)
```

```python
# Run the mediastinum segmentation task
from raidionicsrads.compute import run_rads
run_rads(rads_config_filename)
```

```python
import nibabel as nib
import numpy as np
import matplotlib
import matplotlib.pyplot as plt

logging.getLogger().setLevel(logging.WARNING)
data_filename = os.path.join(patient_dir, 'patient-UnitTest3-Mediastinum', 'inputs', 'T0', '1_CT_HR.nii.gz')
lungs_predictions_filename = os.path.join(test_results_dir, 'T0', '1_CT_HR_annotation-Lungs.nii.gz')
airways_predictions_filename = os.path.join(test_results_dir, 'T0', '1_CT_HR_annotation-Airways.nii.gz')
tumor_predictions_filename = os.path.join(test_results_dir, 'T0', '1_CT_HR_annotation-Tumor.nii.gz')
lymphnodes_predictions_filename = os.path.join(test_results_dir, 'T0', '1_CT_HR_annotation-LymphNodes.nii.gz')
heart_predictions_filename = os.path.join(test_results_dir, 'T0', '1_CT_HR_annotation-Heart.nii.gz')
pulmtrunk_predictions_filename = os.path.join(test_results_dir, 'T0', '1_CT_HR_annotation-PulmonaryTrunk.nii.gz')
pulmveins_predictions_filename = os.path.join(test_results_dir, 'T0', '1_CT_HR_annotation-PulmonaryVeins.nii.gz')
spine_predictions_filename = os.path.join(test_results_dir, 'T0', '1_CT_HR_annotation-Spine.nii.gz')
venacava_predictions_filename = os.path.join(test_results_dir, 'T0', '1_CT_HR_annotation-VenaCava.nii.gz')
ascaorta_predictions_filename = os.path.join(test_results_dir, 'T0', '1_CT_HR_annotation-AscendingAorta.nii.gz')
descaorta_predictions_filename = os.path.join(test_results_dir, 'T0', '1_CT_HR_annotation-DescendingAorta.nii.gz')
aortarch_predictions_filename = os.path.join(test_results_dir, 'T0', '1_CT_HR_annotation-AorticArch.nii.gz')

data = nib.load(data_filename).get_fdata()[:]
lungs_predictions = nib.load(lungs_predictions_filename).get_fdata()[:]
airways_predictions = nib.load(airways_predictions_filename).get_fdata()[:]
tumor_predictions = nib.load(tumor_predictions_filename).get_fdata()[:]
lymphnodes_predictions = nib.load(lymphnodes_predictions_filename).get_fdata()[:]
heart_predictions = nib.load(heart_predictions_filename).get_fdata()[:]
pulmtrunk_predictions = nib.load(pulmtrunk_predictions_filename).get_fdata()[:]
pulmveins_predictions = nib.load(pulmveins_predictions_filename).get_fdata()[:]
spine_predictions = nib.load(spine_predictions_filename).get_fdata()[:]
venacava_predictions = nib.load(venacava_predictions_filename).get_fdata()[:]
ascaorta_predictions = nib.load(ascaorta_predictions_filename).get_fdata()[:]
descaorta_predictions = nib.load(descaorta_predictions_filename).get_fdata()[:]
aortarch_predictions = nib.load(aortarch_predictions_filename).get_fdata()[:]

all_predictions = np.zeros(data.shape).astype('uint8')
all_predictions[lungs_predictions >= 0.5] = 1
all_predictions[airways_predictions >= 0.5] = 2
all_predictions[lymphnodes_predictions >= 0.5] = 3
all_predictions[heart_predictions >= 0.5] = 4
all_predictions[pulmtrunk_predictions >= 0.5] = 5
all_predictions[pulmveins_predictions >= 0.5] = 6
all_predictions[venacava_predictions >= 0.5] = 7
all_predictions[spine_predictions >= 0.5] = 8
all_predictions[ascaorta_predictions >= 0.5] = 9
all_predictions[descaorta_predictions >= 0.5] = 9
all_predictions[aortarch_predictions >= 0.5] = 9
all_predictions[tumor_predictions >= 0.5] = 10

# Plotting corresponding inputs and predictions for a single 2D slice
colors = ['black','green','blue','red', 'purple', 'orange', 'brown', 'pink', 'olive', 'cyan', 'plum']
cmap = matplotlib.colors.ListedColormap(colors)

slice_nr = 472
fig, ax = plt.subplots(3, 3, figsize=(20, 20))
ax[0, 0].imshow(np.rot90(data[..., slice_nr], k=3), cmap="gray")
ax[0, 1].imshow(np.rot90(all_predictions[..., slice_nr], k=3), cmap=cmap)
ax[0, 2].imshow(np.rot90(lungs_predictions[..., slice_nr], k=3), cmap="Reds")
ax[1, 0].imshow(np.rot90(airways_predictions[..., slice_nr], k=3), cmap="Reds")
ax[1, 1].imshow(np.rot90(pulmtrunk_predictions[..., slice_nr], k=3), cmap="Reds")
ax[1, 2].imshow(np.rot90(pulmveins_predictions[..., slice_nr], k=3), cmap="Reds")
ax[2, 0].imshow(np.rot90(lymphnodes_predictions[..., slice_nr], k=3), cmap="Reds")
ax[2, 1].imshow(np.rot90(venacava_predictions[..., slice_nr], k=3), cmap="Reds")
ax[2, 2].imshow(np.rot90(spine_predictions[..., slice_nr], k=3), cmap="Reds")


for i, title in enumerate(["Data", "All labels", "Lungs", "Airways", "Pulmonary trunk", "Pulmonary veins", "Lymph Nodes", "Vena cava", "Spine"]):
  ax[int(i/3), i%3].set_title(title)
plt.show()
```
