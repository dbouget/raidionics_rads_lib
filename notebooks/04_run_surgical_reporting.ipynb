```python
# Download the pipeline
import os
import json

pipeline_dir = os.path.join(test_dir, 'pipelines')
os.makedirs(pipeline_dir, exist_ok=True)

pipeline_json = {}
step_index = 1
step_str = str(step_index)
pipeline_json[step_str] = {}
pipeline_json[step_str]["task"] = "Classification"
pipeline_json[step_str]["inputs"] = {}  # Empty input means running it on all existing data for the patient
pipeline_json[step_str]["target"] = ["MRSequence"]
pipeline_json[step_str]["model"] = "MRI_SequenceClassifier"
pipeline_json[step_str]["description"] = "Classification of the MRI sequence type for all input scans."

# Defining the models to run on the pre-operative data (timestamp 0)
step_index = step_index + 1
step_str = str(step_index)
pipeline_json[step_str] = {}
pipeline_json[step_str]["task"] = 'Model selection'
pipeline_json[step_str]["model"] = 'MRI_TumorCore'
pipeline_json[step_str]["timestamp"] = 0
pipeline_json[step_str]["format"] = "thresholding"
pipeline_json[step_str]["description"] = "Identifying the best tumor core segmentation model for existing inputs"

step_index = step_index + 1
step_str = str(step_index)
pipeline_json[step_str] = {}
pipeline_json[step_str]["task"] = 'Model selection'
pipeline_json[step_str]["model"] = 'MRI_Necrosis'
pipeline_json[step_str]["timestamp"] = 0
pipeline_json[step_str]["format"] = "thresholding"
pipeline_json[step_str]["description"] = "Identifying the best necrosis segmentation model for existing inputs"

step_index = step_index + 1
step_str = str(step_index)
pipeline_json[step_str] = {}
pipeline_json[step_str]["task"] = 'Reporting selection'
pipeline_json[step_str]["scope"] = 'standalone'
pipeline_json[step_str]["timestamps"] = [0]
pipeline_json[step_str]["tumor_type"] = "contrast-enhancing"
pipeline_json[step_str]["description"] = "Identifying the reporting method for existing inputs"

# Defining the models to run on the early post-operative data (timestamp 1)
step_index = step_index + 1
step_str = str(step_index)
pipeline_json[step_str] = {}
pipeline_json[step_str]["task"] = 'Model selection'
pipeline_json[step_str]["model"] = 'MRI_TumorCE_Postop'
pipeline_json[step_str]["timestamp"] = 1
pipeline_json[step_str]["format"] = "thresholding"
pipeline_json[step_str]["description"] = "Identifying the best rest enhancing tumor segmentation model for existing inputs"

step_index = step_index + 1
step_str = str(step_index)
pipeline_json[step_str] = {}
pipeline_json[step_str]["task"] = 'Model selection'
pipeline_json[step_str]["model"] = 'MRI_FLAIRChanges'
pipeline_json[step_str]["timestamp"] = 1
pipeline_json[step_str]["format"] = "thresholding"
pipeline_json[step_str]["description"] = "Identifying the best FLAIR changes segmentation model for existing inputs"

step_index = step_index + 1
step_str = str(step_index)
pipeline_json[step_str] = {}
pipeline_json[step_str]["task"] = 'Model selection'
pipeline_json[step_str]["model"] = 'MRI_Cavity'
pipeline_json[step_str]["timestamp"] = 1
pipeline_json[step_str]["format"] = "thresholding"
pipeline_json[step_str]["description"] = "Identifying the best cavity segmentation model for existing inputs"

step_index = step_index + 1
step_str = str(step_index)
pipeline_json[step_str] = {}
pipeline_json[step_str]["task"] = 'Reporting selection'
pipeline_json[step_str]["scope"] = 'standalone'
pipeline_json[step_str]["timestamps"] = [1]
pipeline_json[step_str]["tumor_type"] = "contrast-enhancing"
pipeline_json[step_str]["description"] = "Identifying the reporting method for existing inputs"

# Computing surgical reporting parameters
step_index = step_index + 1
step_str = str(step_index)
pipeline_json[step_str] = {}
pipeline_json[step_str]["task"] = 'Reporting selection'
pipeline_json[step_str]["scope"] = 'surgical'
pipeline_json[step_str]["timestamps"] = [0, 1]
pipeline_json[step_str]["tumor_type"] = "contrast-enhancing"
pipeline_json[step_str]["description"] = "Identifying the reporting method for existing inputs"

print(json.dumps(pipeline_json, indent=4))
with open(os.path.join(pipeline_dir, 'pipeline_surgical_reporting.json'), 'w', newline='\n') as outfile:
    json.dump(pipeline_json, outfile, indent=4, sort_keys=True)
```

```python
# Prepare the configuration file
import configparser
import json
import logging

test_results_dir = os.path.join(results_dir, "output_surgical_reporting")
os.makedirs(test_results_dir, exist_ok=True)

logging.basicConfig()
logging.getLogger().setLevel(logging.DEBUG)
rads_config = configparser.ConfigParser()
rads_config.add_section('Default')
rads_config.set('Default', 'task', 'neuro_diagnosis')
rads_config.set('Default', 'caller', '')
rads_config.add_section('System')
rads_config.set('System', 'gpu_id', "-1")
rads_config.set('System', 'input_folder', os.path.join(patient_dir, "patient-UnitTest2", "inputs"))
rads_config.set('System', 'output_folder', test_results_dir)
rads_config.set('System', 'model_folder', models_dir)
rads_config.set('System', 'pipeline_filename', os.path.join(pipeline_dir, 'pipeline_surgical_reporting.json'))
rads_config.add_section('Runtime')
rads_config.set('Runtime', 'reconstruction_method', 'thresholding')
rads_config.set('Runtime', 'reconstruction_order', 'resample_first')
rads_config.set('Runtime', 'use_stripped_data', 'True')
rads_config.set('Runtime', 'use_registered_data', 'False')

rads_config_filename = os.path.join(test_results_dir, 'rads_config.ini')
with open(rads_config_filename, 'w') as outfile:
    rads_config.write(outfile)
```

```python
# Run the surgical reporting task
from raidionicsrads.compute import run_rads
run_rads(rads_config_filename)
```

```python
# Inspecting the results
import nibabel as nib
import matplotlib
import matplotlib.pyplot as plt
import numpy as np

logging.getLogger().setLevel(logging.WARNING)

# Identifying the files (in their native spaces)
preop_t1_filename = os.path.join(patient_dir, 'patient-UnitTest2', 'inputs', 'T0', 'preop_t1gd.nii.gz')
postop_t1_filename = os.path.join(patient_dir, 'patient-UnitTest2', 'inputs', 'T1', 'postop_t1gd.nii.gz')
preop_tumor_predictions_filename = os.path.join(test_results_dir, 'T0', 'preop_t1gd_annotation-Tumor_MRI_TumorCore.nii.gz')
preop_necrosis_predictions_filename = os.path.join(test_results_dir, 'T0', 'preop_t1gd_annotation-Necrosis_MRI_Necrosis.nii.gz')
postop_resttumor_predictions_filename = os.path.join(test_results_dir, 'T1', 'postop_t1gd_annotation-TumorCE_MRI_TumorCE_Postop.nii.gz')
postop_cavity_predictions_filename = os.path.join(test_results_dir, 'T1', 'postop_t1gd_annotation-Cavity_MRI_Cavity.nii.gz')

# Identifying the segmentations from other spaces inside the T1c space (for viewing purposes)
postop_reg_folder = os.path.join(test_results_dir, "T1", "T1_T1c_space")
flairchanges_t1c_space_fn = None
for _, _, files in os.walk(postop_reg_folder):
    for f in files:
        if "postop_flair_label_FLAIRChanges_registered" in f and "postop_t1gd" in f:
            flairchanges_t1c_space_fn = os.path.join(postop_reg_folder, f)

# Loading the images 
preop_t1_nib = nib.load(preop_t1_filename)
postop_t1_nib = nib.load(postop_t1_filename)
preop_t1 = preop_t1_nib.get_fdata()[:]
postop_t1 = postop_t1_nib.get_fdata()[:]

# Loading the preoperative segmentations
preop_tumor_pred_nib = nib.load(preop_tumor_predictions_filename)
preop_necrosis_pred_nib = nib.load(preop_necrosis_predictions_filename)
preop_tumor_pred = preop_tumor_pred_nib.get_fdata()[:].astype('uint8')
preop_necrosis_pred = preop_necrosis_pred_nib.get_fdata()[:].astype('uint8')

# Loading the postoperative segmentations
postop_resttumor_pred_nib = nib.load(postop_resttumor_predictions_filename)
postop_cavity_pred_nib = nib.load(postop_cavity_predictions_filename)
postop_flairchanges_pred_nib = nib.load(flairchanges_t1c_space_fn)
postop_resttumor_pred = postop_resttumor_pred_nib.get_fdata()[:].astype('uint8')
postop_cavity_pred = postop_cavity_pred_nib.get_fdata()[:].astype('uint8')
postop_flairchanges_pred = postop_flairchanges_pred_nib.get_fdata()[:].astype('uint8')

# Combining the results
preop_all = np.zeros(preop_t1.shape, dtype='uint8')
preop_all[preop_tumor_pred == 1] = 1
preop_all[preop_necrosis_pred == 1] = 2

postop_all = np.zeros(postop_t1.shape, dtype='uint8')
postop_all[postop_flairchanges_pred == 1] = 3
postop_all[postop_cavity_pred == 1] = 2
postop_all[postop_resttumor_pred == 1] = 1

# Plotting corresponding inputs and predictions for a single 2D slice
colors = ['black','red','blue','green']
cmap = matplotlib.colors.ListedColormap(colors)

fig, ax = plt.subplots(2, 2, figsize=(15, 15))
ax[0, 0].imshow(np.fliplr(np.rot90(preop_t1[:, 72, :], k=3)), cmap="gray")
ax[0, 1].imshow(np.fliplr(np.rot90(preop_all[:, 72, :], k=3)), cmap=matplotlib.colors.ListedColormap(['black','red','blue']))
ax[1, 0].imshow(postop_t1[:, 70, :], cmap="gray")
ax[1, 1].imshow(postop_all[:, 70, :], cmap=matplotlib.colors.ListedColormap(['black','red','yellow', 'green']))

for i, title in enumerate(["Preop T1-CE", "Preop predictions (TumorCore - Red, Necrosis - Blue)", "Postop T1-CE", "Postop predictions (ET - Red, Cavity - Yellow, SNFH- Green)"]):
  ax[int(i/2), i%2].set_title(title)
plt.show()
```
